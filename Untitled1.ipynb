```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.patches as patches

fig, ax = plt.subplots(
    1, 1,
    figsize=(6, 6),
    dpi=150,
    subplot_kw={'projection': ccrs.PlateCarree()}
)

# Continentes en gris claro
ax.add_feature(
    cfeature.LAND,
    facecolor='lightgray',
    edgecolor='none',
    alpha=0.6
)

# Costas y fronteras
ax.coastlines(linewidth=0.7, alpha=0.8)
ax.add_feature(cfeature.BORDERS, linestyle=':', linewidth=0.9, alpha=0.7)

# Extensión
ax.set_extent([233, 292, 10, 40], crs=ccrs.PlateCarree())

# Rectángulo de interés
rect = patches.Rectangle(
    (255, 22),
    262 - 255,
    27 - 22,
    linewidth=.6,
    edgecolor='red',
    facecolor='none',
    transform=ccrs.PlateCarree(),
    zorder=10
)
ax.add_patch(rect)

# Gridlines
gl = ax.gridlines(
    draw_labels=True,
    dms=True,
    x_inline=False,
    y_inline=False,
    color='gray',
    alpha=0.7,
    linestyle='--',
    linewidth=0.5
)
gl.top_labels = False
gl.right_labels = False

ax.tick_params(labelsize=12)

plt.savefig("figsok/mapa_mexico_rectangulo.png", bbox_inches='tight')
plt.show()
```
