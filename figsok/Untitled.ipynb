```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.patches as patches

# Crear figura y ejes con proyección
fig = plt.figure(figsize=(8, 6))
ax1 = plt.axes(projection=ccrs.PlateCarree())

# Establecer la extensión del mapa: [lon_min, lon_max, lat_min, lat_max]
ax1.set_extent([233, 292, 10, 40], crs=ccrs.PlateCarree())

# Agregar características geográficas
ax1.add_feature(cfeature.COASTLINE)
ax1.add_feature(cfeature.BORDERS, linestyle=':')
ax1.add_feature(cfeature.LAND, facecolor='lightgray')
ax1.add_feature(cfeature.OCEAN, facecolor='lightblue')
ax1.gridlines(draw_labels=True)

# Crear y agregar el rectángulo
rect = patches.Rectangle(
    (255, 22),  # Esquina inferior izquierda (lon, lat)
    262 - 255,  # Ancho (longitud)
    27 - 22,    # Alto (latitud)
    linewidth=0.7,
    edgecolor='black',
    facecolor='none',
    transform=ccrs.PlateCarree(),
    zorder=10
)
ax1.add_patch(rect)

# Mostrar el mapa
plt.title("Mapa con extensión definida y rectángulo dibujado")
plt.show()
```
